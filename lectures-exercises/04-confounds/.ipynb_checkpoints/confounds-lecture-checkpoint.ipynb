# Log-Normal Distribution 
* A continuous probability distribution of a random variable whose natural logarithm is normally distributed.
    * $\beta \sim \textrm{Log-Normal}(0,1)\quad$ means that $\log \beta$ is normally distributed,  
      so $\ln\beta\sim\textrm{Normal}(0,1)$

<br/>

* The __PDF__:
$$\textrm{Log-Normal}(x \mid \mu, \sigma) \quad = \quad {1\over x\sigma\sqrt{2\pi}}\exp\,\left[-{1\over 2}\left({\ln x - \mu\strut\over \sigma}\right)^2\right] \quad\quad \text{for } x \gt 0$$

<br/>

* Compare this with the PDF function of a Normal distribution:

  $$\textrm{Normal}(x \mid \mu, \sigma) \quad = \quad \textrm{Normal}(\mu, \sigma)(x) \quad = \quad {1\over \sigma\sqrt{2\pi}}\exp\,\left[-{1\over2}\left({x - \mu\strut\over\sigma}\right)^2\right] \quad\quad \text{for } x \in \mathbb{R}$$
  
* The $x$ in the first term comes from doing the subsitution on a CDF and differentiating to obtain a PDF.

* A __maximum entropy distribution__ for a random variable $X$, for which the mean and variance of $\ln X$ are known, _and we hold no further beliefs_.

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

```python
fig, ax = plt.subplots(1, 1, figsize=[9, 6])
y = np.linspace(-2.0, 2.0, 50)
x = np.linspace(0, 3, 50)

ax.plot(stats.norm.pdf(y, loc = 0, scale = 1), y, 'k:', lw=1.2, alpha=1, label='x = Norm(y|0,1)')
ax.plot(x[3:], np.log(x[3:]), 'k-', lw=1.2, alpha=1, label='y = ln x')

x0 = 0.75

ax.plot(x, stats.lognorm.pdf(x, s = 1.0, scale = 1), 'C1--', lw=1.2, alpha=1, label='lognorm pdf mu=0, sigma=1')

plt.axhline(y = 0, lw=0.8, color = "black", linestyle = "-")
plt.axvline(x = 0, lw=0.8, color = "black", linestyle = "-")
plt.axvline(x = x0, lw=0.8, color = "C1", linestyle = ":")
plt.axhline(y = np.log(x0), lw=0.8, color = "C1", linestyle = ":")
plt.annotate("a", (x0 + 0.05,0.05))
plt.annotate("b", (0.05,np.log(x0)+0.05))
plt.annotate("c", (stats.norm.pdf(np.log(x0), loc = 0, scale=1)+0.05, np.log(x0)+0.05))
plt.annotate("d", (x0 + 0.05, stats.lognorm.pdf(x0, s = 1, scale=1)+0.05))
plt.annotate("x", (3.0,-0.1))
plt.annotate("y", (0.02, 2.05))

plt.annotate("(0,0)", (-0.31,-0.17))
ax.set_ylim(-1.75, 2.15)
ax.set_xlim(-0.1, 2.9)
ax.legend()
plt.show()

# The lognorm API docs:
# https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.lognorm.html
```

* The length of segment $bc$ is not equal to $ad$ but you can intuit that it is (more precisely the fields under the curves should be).
* The discrepancy follows from the 'additional' $x$ in the PDF above 
* They would be equal if the lines plotted where __Cumulative Distribution Functions__ (CDFs) not Probability Density Functions (PDFs).

This plot animates the concept of lognormal (cheating only slightly). Imagine that we have normaly distributed probability mass on x axis (the gray dotted line). We are now going to transform x-es using a log function, so all the negative values land in \[0;1], so they are highly compressed, and all the positive values are just slightly compressed from \[0;+infinity] to \[1;+infinity]. While we are transforming x-es, the probability mass moves a bit along. Now half of the mass should be below 1.0, so we get a "bulge" that was higher than the normal. The right hand side is affected much less as there was more space there.

```python
frames = 100   # Number of frames in the animation
n_points = 100  # Number of points to plot for the distributions

ybot = -.9
height = 2.2

xleft = -2.0
width =  5.7

# Points for the distributions
x = np.linspace(-5.0, 5, n_points)
x_lognorm = np.exp(x)

y_normal = stats.norm.pdf(x)
y_lognorm = stats.lognorm.pdf(x_lognorm, s = 1.0, scale = 1.0)
y_log = np.log(x_lognorm)

fig, ax = plt.subplots(figsize = (width * 1.98, height * 2.75))
ax.plot(x_lognorm, y_log, lw = 1.0, color = 'C1', linestyle = "-", label = "y = ln x")
ax.plot(x, y_normal, lw = 1.0, color = 'gray', linestyle = ":", label = "x ~ norm(μ=0, σ=1)")
ax.plot(x_lognorm, y_lognorm, lw = 1.0, color = 'gray', linestyle = "--", label = "x ~ lognorm(μ=0, σ=1)")
# ax.axvline(x = 0, lw = 0.5, color = "k", linestyle = "-")
ax.arrow(x=0, y=ybot,  dx=0, dy = height, width = 0.00001, length_includes_head = True, head_width = 0.03, head_length = 0.07, fc='k', ec='k')
ax.arrow(y=0, x=xleft, dy=0, dx = width,  width = 0.00001, length_includes_head = True, head_width = 0.023, head_length = 0.08, fc='k', ec='k')
ax.axvline(x = 1, lw = 0.5, color = "k", linestyle = "-", alpha = 0.3)
ax.axhline(y = 0, lw = 0.5, color = "k", linestyle = "-")
ax.set_xlim(xleft, xleft + width)
ax.set_ylim(ybot, ybot + height)
plt.annotate("x", (xleft + width, 0.0), xytext=(xleft + width -0.2, -0.10))
plt.annotate("y", (0.0, ybot + height), xytext=(0.1, ybot + height - 0.10))
plt.annotate("(0,0)", (0.0, 0.0), xytext=(-0.32, -0.10))
plt.annotate("(1,0)", (1.0, 0.0), xytext=( 1.05, -0.10))

nmode = 0
nmode_d = stats.norm.pdf(0.0)
plt.annotate(f"({nmode:.1f},{nmode_d:.1f})", (nmode, nmode_d), xytext = (nmode - 0.5, nmode_d + 0.05))
ax.plot(nmode, nmode_d, 'o', markersize = 3, color='k') 

lnmode = np.exp(0.0 - 1.0)
lnmode_d = stats.lognorm.pdf(lnmode, s = 1.0, scale = 1.0)
plt.annotate(f"({lnmode:.2f},{lnmode_d:.2f})", (lnmode, lnmode_d), xytext = (lnmode - 0.25, lnmode_d + 0.05))
ax.plot(lnmode, lnmode_d, 'o', markersize = 3, color='k') 
ax.legend()

# The interpolation state ranges in (0 = normal, 1 = log-normal)
# The interpolation paramter α controls the frames between 0.0 and 1.0
# but we stretch it to range from -δl to 1 + 2δl fixing the starting and 
# ending frame at the edges so that there is time to appreciate the initial 
# and the final state.
δl = 0.18
δr = 1.0 + 3.0 * δl
α = -δl
step = (1.0/frames) * δr

# initialize the line we will be mutating
line, = ax.plot(x, y_normal, lw = 1.0, color = 'k', linestyle = "-")

def update(frame):
    global α   
    α = α + step
    if α > 1.0 and α <= 1.01:
        α = 1.0 # make sure that we arrive at the final state, despite discretization
    if α > 0.0 and α <= 1.0: 
        x_interp = (1 - α) * x        + α * x_lognorm
        y_interp = (1 - α) * y_normal + α * y_lognorm
        line.set_data(x_interp, y_interp)
    return line,

ani = matplotlib.animation.FuncAnimation(
    fig, update, frames = frames, blit = True, interval = 100)
plt.close(fig)  # hide the static plot
# ani.save('lognormal.gif', writer='pillow', fps=20)
IPython.display.HTML(ani.to_html5_video())
```

# Multicolinearity (Section 6.1)

* A very __strong association__ between two or more predictors


* Consequence: 
  - Introduces a lot of __freedom in the regression__
  - We can take any part from one predictor and complement from the other
  - As a result we get very __high variance on posteriors__ for the parameters, broad useless HDIs


* Multicolinearity __does not harm prediction__
  - The model can be excellent from this point of view
  - It makes model not useful for __explanation / understanding what is going on__.

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# Multicolinearity: An Example 

* Predict height of a person from lengths of their legs. Is it better to know length of one leg or both legs?

<div style="text-align:center">
<img src="fig/fig-00.png" width="20%"/>
</div>

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* Let's create an __artificial data__ set
* Artificial data sets are a very good method to __understand__ stats problems, as we exactly __know all the relations__ in the data

```python
N = 100  # number of individuals
height = np.random.normal(10, 2, N)        # sim total height of each
leg_prop = np.random.uniform(0.4, 0.5, N)  # leg as proportion of height
leg_left = leg_prop * height + np.random.normal(0, 0.02, N)   # sim left leg as proportion + error
leg_right = leg_prop * height + np.random.normal(0, 0.02, N)  # sim right leg as proportion + error

# Let's put it all into a DataSet object
d = xr.Dataset(
    {
        "height": ("index", height),
        "leg_left": ("index", leg_left),
        "leg_right": ("index", leg_right)
    },
    coords={ "index": np.arange(N) })
d
```

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

We now want to regress the height from the lengths of the legs.  
Let's use both legs (both variables) to predict height.

```python
with pm.Model() as m_6_1:
    a  = pm.Normal("a", 10, 100)
    bl = pm.Normal("bl", 2, 10)
    br = pm.Normal("br", 2, 10)

    # Not using MutableData nodes, as I wont be doing posterior predictions or counterfactual plots
    # mu = pm.Deterministic("mu", a + bl * d.leg_left.values + br * d.leg_right.values)
    sigma = pm.Exponential("sigma", 1)

    height = pm.Normal("height", mu = a + bl * d.leg_left.values + br * d.leg_right.values, sigma = sigma, 
                       observed = d.height.values)

    m_6_1_idata = pm.sample(3000, random_seed = rng)
```

* Before we look at the summary, 
  - let's __write the model in mathematical notation__
  - __what values of a, bl, br__ do we expect?

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

```python
az.summary(m_6_1_idata.posterior, round_to=2, kind="stats")
```

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

```python
az.plot_forest(
    data = m_6_1_idata.posterior, 
    colors = ["k"],
    kind = 'forestplot',
    # kind = 'ridgeplot',  # try ridgeplot instead for more information 
    hdi_prob=0.94, 
    combined = True, # plots the result for all four chains combined
    ridgeplot_alpha = .6,
    figsize = (6,2), 
    textsize = 12);
```

* What is going on? It turns out that __many values__ for both bl and br are __possible__ in the prior
* Even though the combinations are not arbitrary, we cannot see this in univariate posterior plots, like above
* Let's plot the bivariate JPD for bl and br:

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

```python
fig, [ax1, ax2] = plt.subplots(1, 2, figsize=[7, 3])

br_ = m_6_1_idata.posterior.br[0]
bl_ = m_6_1_idata.posterior.bl[0]

# code 6.5
ax1.scatter(br_, bl_, color = "k", alpha=0.05, s=20)
ax1.set_xlabel("br")
ax1.set_ylabel("bl")

# code 6.6
az.plot_kde((br_ + bl_).values, plot_kwargs={'color': 'k'})
ax2.set_ylabel("Density")
ax2.set_xlabel("sum of bl and br");
plt.show()
```

* The posterior for the sum of bl and br is also less flat, and it has the mode slightly above 2, as expected

* When several parameters (features) are strongly correlated, the model retains the prediction ability
  - but __we cannot use the posterior to explain__ which features corelate with the outcome  

* We might __wrongly conclude__ that none of the parameters is a good predictor,
  - even though __either is a good predictor__.

* This is not an issue, if we are not interested in correlation and causation. Unfortunately, we often are, especially in scientific inference.


<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# Post-Treatment Bias (Section 6.2)

* People often worry about confounders that are not in the model


* Unfortunately, __adding too many variables__ in the model can also be __harmful__


* With __multicolinearity__ adding variable __obscured conclusions__.


* __Post-treatment bias__ can lead to __entirely wrong conclusions__.


* Post-treatment bias appears when you use as a __regressor that is not directly applied (treatment) but is a result of your treatment__
  - it confuses causality in the model

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# Post-Treatment Bias Example


* We grow plants, and measure their height under different fungal soil treatments


* We measure: 
    - initial height (**h0**), 
    - final height (**h1**), 
    - treatment (**T**), 
    - presence of fungus (**F**)


* Let us fake this experiment (__Note Again:__ how useful it is to fake experiments before running them!)

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

```python
# number of plants
N = 100
# simulate initial heights
h0 = np.random.normal(10, 2, N)
# assign treatments and simulate fungus and growth
treatment = np.repeat([0, 1], N / 2)
# fungus develops with 50% chance, but only 10% under treatment
fungus = np.random.binomial(n=1, p=0.5 - treatment * 0.4, size = N)
h1 = h0 + np.random.normal(5 - 3 * fungus, size = N)

# Let's put it all into a DataSet object
d = xr.Dataset(
    {
        "h0": ("index", h0),
        "h1": ("index", h1),
        "treatment": ("index", treatment),
        "fungus": ("index", fungus)
    },
    coords={ "index": np.arange(N) })
d
```

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

```python
az.summary(d.to_dataframe().to_dict(orient="list"), kind="stats", round_to=2)
```

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

**(skip in class)**
* We use the following model first for simplicity (let's ust try to see what a typical growth is across the spectrum):

$$
\begin{align}
h_{1,i} & \sim \textrm{Normal}(\mu_i, \sigma) \\
\mu_1   & = h_{0,i} \cdot p \\
p       & \sim \textrm{Log-Normal}(1.0, 0.25)
\end{align}
$$

* Let's use a log-normal prior for p, which gives as typical growth from $0.56$ to $1.5$

```python
sim_p = np.random.lognormal(0, 0.25, int(1e4))
# Note: scipy/np use a different parameterization of lognormal 
# (https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.lognorm.html)

az.summary(sim_p, kind="stats", round_to=2)
```

(x) is $p$ above.  Just the numpy table carries no variable name, so arviz inserted x by default.

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

Let's try to fit this model:

```python
with pm.Model() as m_6_6:
    p = pm.Lognormal("p", 0, 0.25)

    mu = p * d.h0.values
    sigma = pm.Exponential("sigma", 1)

    h1 = pm.Normal("h1", mu=mu, sigma=sigma, observed=d.h1.values)

    m_6_6_idata = pm.sample(random_seed = rng)
```

```python
az.summary(m_6_6_idata.posterior, round_to=2, kind="stats")
```

* We observe 40% growth on average, and a reasonably small spread. 

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

Let's build a model with both the treatment and fungus measurements.

$$
\begin{align}
h_{1,i} & \sim \textrm{Normal}(\mu_i, \sigma) \\
\mu_i   & = h_{0,i}p \\
p       & = \alpha + \beta_T T_i + \beta_F F_i \\
\alpha  & \sim \textrm{Log-Normal}(0, 0.25) \\
\beta_T & \sim \textrm{Normal}(0, 0.5) \\
\beta_F & \sim \textrm{Normal}(0, 0.5) \\
\sigma  & \sim \textrm{Exponential}(1)
\end{align}
$$

```python
with pm.Model() as m_6_7:
    a = pm.Normal("a", 0, 0.2)
    bt = pm.Normal("bt", 0, 0.5)
    bf = pm.Normal("bf", 0, 0.5)

    p = a + bt * d.treatment.values + bf * d.fungus.values

    mu = p * d.h0.values
    sigma = pm.Exponential("sigma", 1)

    h1 = pm.Normal("h1", mu=mu, sigma=sigma, observed=d.h1.values)

    m_6_7_idata = pm.sample(random_seed = rng)
```

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

```python
az.summary(m_6_7_idata.posterior, round_to = 2, kind = "stats")
```

* It appears that our __treatment has no influence on growth__, only the fungus has!
* So one is inclined to publish a research article with negative result: this treatment does not work (or better start to do research on another treatment).
* But this treatment does work most likely!

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* How can that be?

* We work with the following causal DAG:

```python
growth = CausalGraphicalModel(
    nodes=["H0", "H1", "F", "T"],
    edges=[
        ("H0", "H1"), 
        ("F", "H1"), 
        ("T", "F"),
    ]
)

growth.draw()
```

* The final height is influenced by the initial height
* And it is influenced by presence of fungi, which we introduce to the environment by the treatment.
* It turns out that $T$ and $H_1$ are d-separated by $F$

* **def.** Two variables $X$ and $Y$ in a causal DAG are **d-separated** by a set of variables $Z$ iff
* Every undirected path between $X$ and $Y$ is
  - a directed path: $X \rightarrow \dots \rightarrow Y$, or $X \leftarrow \dots \leftarrow Y$, and the path contains variables in $Z$
  - contains a fork, so a middle node $m$ such that the part from $X$ to $m$ is directed and the part from $Y$ to $m$ is a directed path: $X \leftarrow \dots \leftarrow m \rightarrow \dots \rightarrow Y$, $m \in Z$
  - contains a collider (an inverted fork), $X \rightarrow \dots \rightarrow m \leftarrow \dots \leftarrow Y$, such that $m \notin Z$ and no descendant of $m$ is in $Z$.
* "d" stands for "directional"
* In case of $T$ and $H_1$  the first case kicks in (we talk about the other cases below)
* If $X$, $Y$ are d-separated then they are independent of each other given the variables in $Z$

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* Here $H_1$ is d-separated from $T$, when we condition on $F$. 

```python
print(f'Is H1 d-separated from T given F?   {growth.is_d_separated("H1", "T", {"F"})}')
```

```python
growth.get_all_independence_relationships()
```

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* If we know $F$, $T$ can add no additional information: $H_1 \!\perp\!\!\!\perp T \mid F$


* If we know the value of $F$, the value of $T$ that led to it has no remaining influence on $H_1$.


* __Conclusion__: If we want to assess the effect of treatment on $H_1$, then __we should not include $F$ in the regression__ (as we did initially)
* These kinds of decisions cannot be made without a causal perspective (the big-world information)!

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# Collider Bias (Sec 6.3)


* But it can be worse ...


* What if the treatment really does not have influence on the growth, but there is another variable (M)oisture that is hidden, and it influences positively both the growth of the fungus and the plant?


* If this is the case, the model will report that (T)reatment has a positive effect, even if it does not!

* We simulated this model for 1000 years, and then take a snapshot of the population
* I include the simulation code, but we don't discuss it

```python
def inv_logit(x):
    return np.exp(x) / (1 + np.exp(x))
```

```python
def sim_happiness(N_years = 100, seed = 1234):
    np.random.seed(seed)
   
    # Initial population setup
    # the number of persons in the sample
    N = 20 * 65 
    # 20 persons in each age below 65
    age = np.repeat(np.arange(65), 20)
    # each person in each age has a different happiness level between -2 and +2
    happiness = np.repeat(np.linspace(-2, 2, 20), 65)
    married = np.zeros(N, dtype=bool)

    for i in range(N_years):
        # age population
        age = age + 1
        # replace old folk with new folk
        ind = (age == 65)
        age[ind] = 0
        married[ind] = False
        happiness[ind] = np.linspace(-2, 2, 20)
        
        # do the work
        elligible = (married == 0) & (age >= 18)
        marry = np.random.binomial(1, inv_logit(happiness[elligible] - 4)) == 1
        married[elligible] = marry
        
    popn = xr.Dataset({
        "age": ("index", age),
        "happiness": ("index", happiness),
        "married": ("index", married),
    },
    coords={ "index": np.arange(N) })
    
    # Sort the Dataset by 'age'
    sorted_indices = np.argsort(popn.age.values)
    popn_sorted = popn.isel(index=sorted_indices)    
    
    return popn_sorted
```

```python
popn = sim_happiness()

popn_summ = popn.copy()
popn_summ["married"] = popn_summ["married"].astype(
    int
)  # this is necessary before using az.summary, which doesn't work with boolean columns.
az.summary(popn_summ.to_dataframe().to_dict(orient="list"), kind="stats", round_to=2)
```

* People are roughly as happy as unhappy in the population (as we simulated)
* A bit less than 30% are married on average

```python
plt.subplots(1,1, figsize=(9,4))
plt.scatter(popn.where(popn.married == False).age, popn.where(popn.married == False).happiness,
            alpha=0.5, edgecolor='darkgray', s = 20, color= 'none', label='unmarried')
plt.scatter(popn.where(popn.married == True).age, popn.where(popn.married == True).happiness,
            alpha=0.5, edgecolor='none', s = 20, color= 'k', label='married')

plt.xlabel('Age')
plt.ylabel('Happiness')
plt.title('Simulated data')
plt.legend()

plt.show()
```

* If we condition on married folks (include married status in the model), we will conclude that happiness and age are inversely correlated. But there was nothing in our generation method that would cause that.

* Since the collider is a phenomenon not in correlation but in the causal graph, it is an inherent problem of the "big world". 

* We don't need to measure it, or to include it in the model to observe the effects.

* Unfortunately we **often get data** already conditioned on the collider (survival bias in data)

# Summary of Confounder Patterns

<img src="fig/fig-11.png"/>

* __Fork__: the classic hidden confounder variable
  - condition on Z, make X and Y independent, even if they are dependent on a common cause
  - the problem appears if we lack variables in the model; we don't condition on Z, and spurious dependence appears; __include Z__ to expose non-causal paths
  - __the path from X and Y is blocked by conditioning on Z__
  - Example: WaffleHouse-DivorceRate


* __Pipe__: post-treatment bias
  - condition on Z, X becomes independent of Y
  - Appears when we have needless variables in the model, __remove Z__ to expose the causal path
  - __the path from X to Y is blocked by conditioning on Z__
  - Example: treatment-funghi-growth


* __Collider__: X and Y are independent unless you condition on Z.
  - the problem appears if we condition on a collider (perhaps unconciously) and introduce a false dependency from Y or to X (or other variables)
  - one has to be careful if the data we have is not conditioned on a collider already
  - __the path from X to Y is open when conditioning on Z__; __remove Z__ to avoid the non-causal path
  - Example: the modified fungi example, newsworthines-trustworthiness, aga-marriage-happines


* __Descendant__: Any of the above indirectly 
  - Note that the descendant problem can be a variant of the fork, the pipe, or the collider pattern
  - all these problems appear even if we don't condition on Z, but on a variable correlated with it (a descendant D)

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# A procedure to address these 4 confounds, assuming you have a complete DAG





Given a causal DAG, it is always possible to say which, if any, variables one must control for in order to shut all the backdoor paths (non-causal paths), and which variables one must not control for, in order to avoid making new confounds. 

1. List all path connecting $X$ (the potential cause of interest) to $Y$ (the outcome of interest)
2. Classify each path by whether it is open or __closed (contains a collider)__
3. A path is a __backdoor__ if it has an arrow entering $X$
4. Decide which variables to condition on (to include in the model) or not to close the non-causal paths, using the patterns listed above, for forks, pipes, and colliders.

Because the set of rules is finite, there are libraries that can compute it for you, but it is always good to understand the problems yourself from the first principles.

The `CausalGraphicalModels` has two useful functions `is_valid_backdoor_adjustment_set` and `get_all_backdoor_adjustment_sets`.

# An Example
* Consider the following casual graph

<img src="fig/fig-12.png"/>


* We are interested in influence of $X$ on $Y$
* We cannot measure $U$. 
* We can choose to measure $A$, $B$, and $C$. 
* Should we? Which one?

# The Analysis of the example

* We have the following backdoor paths from $X$ to $Y$:
  1. $ X \leftarrow U \leftarrow A\rightarrow C \rightarrow Y $
  2. $ X \leftarrow U \rightarrow B \leftarrow C \rightarrow Y $
* They are both backdoor paths
* Path (1) contains no collider so it is open, 
    - A joint influence on $X$ and $Y$ might be confusing us (like the state location on waffle-house and divorces)
    - We need to introduce variable $A$ (or $C$) in our regression to control for this influence
* Path (2) is closed as it contains a collider
    - if we don't condition on B and open the influence, we are safe wrt to $X$ and $Y$. 
    - Conditioning on $B$ could introduce a collider bias (like in the marriage-happiness example, and others)
    - When we close paths (like 1), we want to make sure we do not open closed paths like (2).
* To close path (1) we can condition on $A$ or $C$
    - $C$ is likely a better choice as it is closer to $Y$ (direct influence will have stronger inference effects than indirect).
