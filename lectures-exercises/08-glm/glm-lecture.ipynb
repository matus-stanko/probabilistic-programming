## Maximum Entropy Distributions


* A distribution with the biggest entropy is the __widest__ and __least informative__ distribution (in a constrained set you consider)
* In other words: the __most uncertain__ distribution in a constrained set you consider.
* Most uncertain means that it adds __no arbitrary information__ besides your constraints
* When building models we want to use maximum entropy (maxent) distributions.

<br/>

* For theoretical reasons
    - A maximum entropy likelihood is the one that satisfies the constraints (what parameters we know), but it is most spread in probability otherwise, so introduces __no additional bias in the parameter likelihood__
    - A maximum entropy prior is the one that satisfies the constraints (what we know about data), but it is most spread in probability otherwise, so introduces __no additional bias in the parameter priors__
* To satisfy reality:
    - __Nature tends to produce empirical distributions of high entropy__ (not necessary Gaussian)
* Fortunately, it turns out that maximum entropy distributions are the familiar distributions from classical probability theory! So we do not need to learn a lot of new esoteric stuff.

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# Definition

* __Entropy__ (discrete and continuous, sometimes called _differential entropy_):

$$ H(p) = - \sum_i p_i \log p_i \quad\quad\quad H(p) = - \int p(x) \log p(x)\, dx$$

* The __maximum entropy distribution__ $p$ is a distribution that 
    1. __Satisfies our constraints__ (for instance a known mean, variance, etc.) 
    2. It __has the largest entropy__ of all distributions that satisfy the constraints

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# The Wallis Derivation 

* Proving the above, instead of experimenting

__Goal__: 
* Prove that in the above process the distribution which can be obtained in most ways is the maximum entropy distribution.

__Assumptions__:
* We have some set of constraints on a distribution
* $M$ buckets, $N$ individual pebbles, each tossed with equal chance into any of the buckets. $N$ is large
* After tossing the stones we convert the sizes of buckets to a discrete distribution: $p_i = n_i/N$
* We repeat this many times, and take a distribution that appeared most times as being the preferred one.

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

__Proof.__

* The probability of obtaining a partial distribution of pebbles in the buckets is a multinomial distribution (a generalization of binomial to more than two outcomes)

$$\Pr(n_1, n_2, \ldots, n_M) = {N!\over n_1!n_2!\dots n_M!}\prod_{i=1}^M\left(1\over M\right)^{n_i} = {N!\over n_1!n_2!\dots n_m!} \left(1\over M\right)^N$$

* $n_i$ is the number of outcomes of type $i$ (the $i$th bucket), $\sum_i^M n_i = N$
* $1/M$ is the probability of outcome of type $i$ (landing in the $i$th bucket)
* $(1/M)^{n_i}$ the probability of $n_i$ outcomes of type $i$
* $N!$ the number of reorderings of experiments that give the same outcome
* The denominator accounts for the number of reorderings within the same bucket, which does not change the outcome. (We don't have pebble identity)
* Let's abbreviate the ugly fraction as $W$, so we get $W\cdot \left(1\over M\right)^N$

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* The distribution appearing most often is the most likely sequence $n_1,\dots, n_M$ in the above pmf, so its __mode__
* Note that $\left(1\over M\right)^N$ is constant (indepentend of the values of $n_1,\dots, n_M$
* So the mode is the sequence $n_1,\dots, n_M$ that maximizes $W$.

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* __Observation__: a sequence maximizees $W$ iff it maximizes ${1\over N}\log W$ 
    - because $\log$ is monotonic and ${1\over N}$ is a constant


* Let us expand ${1\over N}\log W$:
    - using (1) logarithm of a division, (2) logarithm of a multiplication, and (3) $n_i = p_iN$
    
$$ {1\over N} \log W = {1\over N} \log \left[{N!\over n_1!n_2!\dots n_m!}\right] = {1\over N} \left(\log N! - \sum_i \log\left[(Np_i)!\right]\right)$$



<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* Now we use __Stirling's approximation__: for large $N$, $\log N!$ is approximately $N\log N - N$

$$ {1\over N} \log W \approx {1\over N} \left[N\log N - N - \sum_i \left(Np_i\log(Np_i)-Np_i\right)\right] $$
$$ = {1\over N} \left[N\log N - N - N\sum_i p_i(\log N + \log p_i) + N\sum_ip_i\right] $$
$$ = {1\over N} \left[N\log N - N - N\log N\sum_i p_i - N \sum_i p_i \log p_i + N\right] $$
$$ = {1\over N} \left[- N \sum_i p_i \log p_i\right] = - \sum_i p_i \log p_i $$

* So (up to the Stirling approximation) the distribution that has the highest probability as an outcome, is the one that has highest entropy in this experiment, because the mode of entropy will be in the same place as mode of $W$ 

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# Bayes Theorem Gives Us Maximum Entropy Posterior

* The Wallis derivation can be generalized to the case where a chance of landing a pebble in each bucket is __not uniform__, but given by a distribution $q_i$
* $q_i$ is the prior on the distribution of pebbles (in practice it will be a prior on parameters defining the set of distributions)
* One can show that then the mode of the distribution space is the distribution that maximizes the following:

$${1 \over N} \log \Pr(n_1, \dots, n_M) \approx - \sum_i pi \log (p_i/q_i)$$

* which is the negative of KL divergence (relative entropy) of $p_i$ and $q_i$.
* So the resulting distribution minimizes KL divergence (relative entropy, so relative uncertainty with respect to the prior)
* Thus __Bayes theorem returns the posterior distribution that is most like the prior but still consistent with the constraints__ ("most like" as in having the smallest KL-divergence).

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# Gaussian (Section 10.1.1)

* Gaussian is the mexent distribution __given a known mean and variance__ (the constraint constraints the mean and variance)
* Intuitively, the flatter the distribution is, the higher its entropy, so in the figure the flattest distribution should have highest entropy
* But a flatter distribution has a __higher variance__, so entropy maximization and the variance constraint are opposing

* It turns out that the normal distribution is the flattest distribution that has the given variance

# Thm. Gaussian is Maxent, under a mean and variance constraint

* Let $p(x)$ be a __Gaussian density function__ with mean $\mu$ and variance $\sigma^2$ 

    $$p(x) = {1\over \sigma\sqrt{2\pi}} \exp \left[-{1\over 2}\left({x-\mu \over \sigma}\right)^2\right]$$
    

* Let us start with calculating the __entropy of Gaussian__ (McElreath skips this step):

    $$H(p) = - \int_{-\infty}^{+\infty} p(x) \log p(x) dx = $$ 
    
    $$ = -\int_{-\infty}^{+\infty} p(x) \left[ \log\left((2\pi\sigma^2)^{-1/2}\right) - {1\over 2}\left({x-\mu\over \sigma}\right)^2\right]dx = $$
    
    $$ = -\int_{-\infty}^{+\infty} p(x) \left[-{1\over2} \log\left(2\pi\sigma^2\right) - {1\over 2}\left({x-\mu\over \sigma}\right)^2\right]dx = $$
    
    $$ = \int_{-\infty}^{+\infty} p(x) {1\over2} \log\left(2\pi\sigma^2\right)dx + \int_{-\infty}^{+\infty} p(x) {1\over 2}\left({x-\mu\over \sigma}\right)^2dx = $$

    $$ = {1\over2} \log\left(2\pi\sigma^2\right)\int_{-\infty}^{+\infty}  p(x) dx + {1\over 2\sigma^2} \int_{-\infty}^{+\infty} p(x) \left({x-\mu}\right)^2dx = $$
    
    $$ = {1\over2} \log\left(2\pi\sigma^2\right) + {1\over 2\sigma^2} \mathrm{VAR}(x) = $$
        
    $$= {1\over2} \log\left(2\pi\sigma^2\right) + {1\over 2} = {1\over2} \log\left(2\pi\sigma^2\right) + {1\over 2}\log e = {1\over 2} \log (2\pi e\sigma^2)$$
    
* Let us note this result - it will come usefule a bit later.

* Let $q(x)$ be some other pdf with the same variance $\sigma^2$ and also defined over all real numbers 
   - (we can ignore mean, as it does not influence entropy -- although this would require a separate albeit very simple proof)

<br/><br/>

* __A digression__. Let's prove an inequality involving cross-entropy and entropy. Consider relative entropy:

$$D_\mathrm{KL}(q,p) = \int_{-\infty}^{+\infty} q(x) \log \left[{q(x)\over p(x)}\right] dx  $$

$$ = \int_{-\infty}^{+\infty} q(x) \left[\log q(x) - \log p(x)\right]dx $$

$$ = \int_{-\infty}^{+\infty} q(x) \log q(x)dx - \int_{-\infty}^{+\infty} q(x) \log p(x) dx $$

$$ = -H(q) - H(q,p) $$

* $H(q,p)$ is known as cross-entropy

* Since relative entropy (KL-divergence) is always non-negative we know that:

$$ -H(q) - H(q,p) \geq 0 $$

$$ - H(q,p) \geq H(q) $$

* __End of digression.__ 

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>


* Let us now compute $H(q,p)$ for our case of $p$ being normal, and see what does it tell as about $H(q)$ using the above inequality:

$$H(q,p) = \int_{-\infty}^{+\infty} q(x)\log p(x) dx $$

$$ = \int_{-\infty}^{+\infty} q(x) \log\left[{1\over \sigma\sqrt{2\pi}} \exp \left[-{1\over 2}\left({x-\mu \over \sigma}\right)^2\right]\right]dx $$

$$ = \mathrm{E}_q \log\left[{1\over \sigma\sqrt{2\pi}} \exp \left[-{1\over 2}\left({x-\mu \over \sigma}\right)^2\right]\right] $$

$$ = \mathrm{E}_q \log{1\over \sigma\sqrt{2\pi}} + \mathrm{E}_q \left[-{1\over 2}\left({x-\mu \over \sigma}\right)^2\right] $$

$$ = \log{1\over \sigma\sqrt{2\pi}} -{1\over 2\sigma^2}\mathrm{E}_q (x-\mu)^2 $$

$$ = - \log{\sqrt{2\pi\sigma^2}} -{1\over 2\sigma^2}\mathrm{VAR}_q (x) $$ 

$$ = - {1\over 2}\log(2\pi\sigma^2)  -{1\over 2} $$

$$ = - {1\over 2}\log(2\pi\sigma^2)  -{1\over 2}\log e = - {1\over 2} \log(2\pi e\sigma^2) = - H(p) $$

* Recall that the value we obtained for $H(q,p)$ is just a negation of $H(p)$ (see above in the begining of the proof). So

$$ H(p) = -H(q,p) \geq H(q) $$

* We conclude that $p$ has higher entropy than $q$. __q.e.d.__

# Binomial (Section 10.1.2)

* Defines the pmf for __number of successes__ in a $n$ indepdendent Bernoulli experiments with __success chance__ $p$:

$$ \Pr(y\mid n,p) = {n! \over y!(n-y)!} p^y (1-p)^{n-y} $$

* Or in the form that enumerates the obtained path, not the number of succesess (in this formulation $y_i$s are zero or one):

$$ \Pr(y_1, \ldots, y_n \mid n, p) =  p^{\sum y_i} (1-p)^{n-\sum y_i} $$


* We want to prove that this distribution has __higher entropy__ than any distribution that satisfies two constraints

# Thm. Binomial is Maxent 

* More precisely: _the binomial distribution is maximum entropy distribution in the set of generalized binomial distributions with the same expected number of successes_

* Let $p$ be a __binomial__ distribution, $p_i$ describes the chance of getting $x_i$ succesess, number of failures is $n - x_i$
* Let $\lambda = \sum_i p_i x_i$ be the __expected value__ of $p$ (the expected number of successes), then $\lambda/ n = p$ is the chance of success
* Let $q_i$ be some __other distribution__ over the same set of sequences, and the same expected number of successes $\lambda$
* We know from the previous proof that $-H(q,p) \geq H(q)$
* Let's calculate $- H(q,p)$ ($i$ ranges over all possible different sequences of outcomes):

$$-H(q,p) = - \sum_i q_i \log p_i = - \sum_i q_i \log \left[ p^{\sum y_i} (1-p)^{n-\sum y_i} \right] $$

$$ =  - \sum_i q_i \log \left[ \left({\lambda\over n}\right)^{x_i} \left(1-{\lambda\over n}\right)^{n-x_i} \right] $$

$$ = - \sum_i  q_i \left( x_i \log \left[{\lambda\over n}\right] + (n-x_i) \log \left[1-{\lambda\over n}\right]\right) $$

$$ = - \sum_i  q_i \left( x_i \log \left[{\lambda\over n}\right] -x_i \log \left[{n - \lambda\over n}\right] + n \log \left[{n - \lambda\over n}\right]\right) $$

$$= - \sum_i  q_i \left( x_i \log \left[{\lambda\over n - \lambda}\right] + n \log \left[{n - \lambda\over n}\right]\right) $$

$$ = - n \log \left[{n - \lambda\over n}\right] - \log \left[{\lambda\over n - \lambda}\right] \sum_i  q_i x_i $$

* where the last summation is the expected value of $q$. If this expected value is the same for $p$ (so $\lambda$) then

$$ = - n \log \left[{n - \lambda\over n}\right] - \lambda \log \left[{\lambda\over n - \lambda}\right] = H(p) $$

* The last step requires calculating $H(p)$ (an almost identical calculation as for the cross-entropy, above)
* We then get that $H(p) = - H(q,p) \geq H(q)$, concluding the proof.

# Generalized Linear Model (Sec. 10.2)

* So far our models had a classic linear regression form:

$$ y_i \sim \mathrm{Normal} (\mu_i, \sigma) $$
$$ \mu_i = \alpha + \beta x_i $$

* Then if we used posterior mean we optimized expected mean square error, if we chose posterior mode we optimized expected distance from the mean (Euclidian).

* But what if we wanted to optimize another parameter than a mean of a Gaussian?

* We use a __different likelihood__ function, and possibly a different __link function__ 

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* For instance, if we wanted to optimize the probability of success in a Binomial distribution, we could use the following model:

$$ y_i \sim \mathrm{Binomial}(n, p_i) $$

$$ f(p_i) = \alpha + \beta(x_i - \overline x) $$

* __Two differences__ from "Gaussian" regression
    - Essential: we use binomial because it is __maxent for a number of succeses__ from n trials with expected value $np$.
    - "Geocentric/Epicyclic": we need to adjust the linear equation (function $f$) to map all real values that can be produced by a linear model to __fit in the interval__ $[0;1]$, which are legal values for a probability parameter
    - The function $f$ is called a __link function__, and $f^{-1}$ is called an inverse link function

__Exponential__
* __Support__: positive real values
* __Intuition__: represents displacement in time or space
* __Constraints__: the only thing we know is the expected displacement $\lambda^{-1}$ (and the positive support)

__Gamma__
* __Support__: positive real values
* __Intuition__: represents displacement, but with mode which is positive (not-zero like with Exponential), appears as a sum of a few exponentials.
* __Constraints__:
  - the displacement (event) can happen only after a small number of exponentially distributed events happen (so we delay exponentially a bit, we __sum exponentials__, and then our event can happen)
  - we know the __mean__, and average logarithm for the distribution
 
__Poisson__
* __Support__: non-negative integers (like Binomial we count events)
* __Intuition__: a binomial with a very large count $n$ (effectively infinite) and a small probability of success $p$ ($p$ is the only parameter)
* __Constraints__: we know is the succes chance for an event, or more precisely the rate $\lambda$ such that $ p = \lim_{n\rightarrow\infty} {\lambda\over n}$

# Link function

* A link function maps a linear space of a model like $\alpha + \beta(x_i - \overline{x})$ onto the non-linear space of a parameter like $\theta$

* Two most popular link functions are:
    - __log__: maps any positive real value into a value on a real line in a continuous and monotonic fashion
    - __log it__: maps any value in interval $[0;1]$ into a value on a real line in a continuous and monotonic fashion
    

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# The Logit Link (Logistic regression)

<img src="fig/fig-06.png" width=50%
     style="float: right; padding-left: 2cm;" />

* The logit link is popular when working with binomial (or any likelihood, where the parameter is a probability value).

$$y_i \sim \mathrm{Binomial} $$

$$ \mathrm{logit}(p_i) = \alpha + \beta x_i $$

* Note the drawing to the right is "transposed", it plots the value of $\mathrm{logit}$ on the $x$ axis, and the values of $p$ on the $y$ axis. (so again an inverse link function)
* Traditionally the link function is defined as written above, but we often think about the inverse (like drawn)

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* $\mathrm{logit}$ is defined as "logarithm of odds" (ratio of success probability and failure probability):

$$ \mathrm{logit}(p_i) = \log {p_i \over 1-p_i}$$

* The definition of link function is somewhat "backwards", when you need to understand what is $p_i$ you need to solve for it:

$$ \log {p_i \over 1-p_i} = \alpha + \beta x_i $$

$$ p_i = {\exp(\alpha+\beta x_i)\over 1 + \exp(\alpha + \beta x_i)} $$

* The above (inverse link function) is called the logistic function, or the inverse-logit function. 
* Of course, interpreting the regression parameters become more difficult
* **This transformation affects the interpretation of parameter estimates.** Now a unit change of $x_i$ value results in $\beta$-sized change of the log-odds value for $p$. Interpretation becomes difficult.
* Little we can do about this though, if the model inherently captures what we know.

<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

# The Log Link

<img src="fig/fig-07.png" width=50%
     style="float: left; padding-right: 2cm;" />

* The logarithm function translates a set of non-negative values (parameter values) into a real line. The plot shows the inverse

* It is good to map a support of a positive parameter (like standard deviation) into a real line in a regression model

$$ y_i \sim \mathrm{Normal} (\mu, \sigma_i) $$

$$ \log (\sigma_i) = \alpha + \beta x_i $$



<br/><br/><br/><br/><br/><br/><br/><br/><br/><br/>

* The plot to the left is again (transposed) and shows the inverse of a log-link function, so the exponential:

$$ \sigma_i = \exp(\alpha + \beta x_i)$$

* __Interpretation of the regression convoluted__ again: a unit increase of the predictor $x_i$ causes an order of magnitude increase (exponential increase proportional to $\beta$) of the target variable (here $\sigma$).

* When we use a non-linear link it is like if we introduced a non-linear interaction of a parameter with itself, because the impact of its change depends not only on the change, but also on the value of the parameter itself.

# GLM: The Formal Definition

The general formal expression of the GLM is:

\begin{align}
  y &\sim \mbox{pdf}(\mu, [\mbox{parameters}]) \\
    \mu &= f^{-1}(\mbox{lin}(x),[\mbox{parameters}])
\end{align}

</br> 

where 
* $\mbox{lin}(x)$ is a linear function over the predictors $x$,
* $f^{-1}(\cdot)$ is the inverse link function
* $y$ is the data which is distributed over the central tendency $\mu$ according to a probability distribution $\mbox{pdf}$.
